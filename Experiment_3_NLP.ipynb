## Experiment 3: Sentence and Word Tokenisation
**Domain Application:** Terms & Conditions Summarizer (Clause and Token Segmentation using NLTK & spaCy)

### Step 1: Tokenisation Using NLTK
Perform sentence and word tokenization on long, compound legal agreements using NLTK's `sent_tokenize` and `word_tokenize`.

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Sample Terms and Conditions excerpt with abbreviations and complex structure
sample_legal_text = """
Welcome to Amazon Services LLC. Please review our Privacy Notice (e.g., data collection rules, cookies policy, etc.) before using our platform.
If you are under 18 years of age, you may use Amazon Services only with involvement of a parent or guardian. Amazon.com reserves all rights to terminate accounts without prior notice.
"""

# Sentence tokenization with NLTK
nltk_sentences = sent_tokenize(sample_legal_text.strip())

print("--- NLTK Sentence Tokenisation ---")
print(f"Total Sentences Identified: {len(nltk_sentences)}\n")
for idx, sentence in enumerate(nltk_sentences, 1):
    print(f"[Sentence {idx}]: {sentence}")

# Word tokenization on first sentence with NLTK
nltk_words = word_tokenize(nltk_sentences[0])
print("\n--- NLTK Word Tokenisation (Sentence 1) ---")
print(f"Total Tokens: {len(nltk_words)}")
print("Tokens List:", nltk_words)
```

### Step 2: Tokenisation Using spaCy
Use spaCy's linguistic model (`en_core_web_sm`) for rule-and-dependency-based sentence segmentation and token attribute inspection.

```python
import spacy

# Load spaCy English language model
nlp = spacy.load("en_core_web_sm")

# Process legal text through spaCy pipeline
doc = nlp(sample_legal_text.strip())

spacy_sentences = [sent.text.strip() for sent in doc.sents]

print("--- spaCy Sentence Segmentation ---")
print(f"Total Sentences Identified: {len(spacy_sentences)}\n")
for idx, sentence in enumerate(spacy_sentences, 1):
    print(f"[Sentence {idx}]: {sentence}")

# Token level inspection with attributes
print("\n--- spaCy Token Level Details (Sentence 1) ---")
first_sent_doc = nlp(spacy_sentences[0])
token_data = []
for token in first_sent_doc:
    token_data.append({
        "Token": token.text,
        "Is_Alpha": token.is_alpha,
        "Is_Punct": token.is_punct,
        "Is_StopWord": token.is_stop,
        "Like_Num": token.like_num
    })

import pandas as pd
df_tokens = pd.DataFrame(token_data)
display(df_tokens.head(12))
```

### Step 3: Comparative Analysis: Python `split()` vs NLTK vs spaCy on Legal Text
Compare token handling across different approaches, focusing on legal contractions (e.g. `can't`, `user's`), abbreviations (`e.g.`, `LLC.`), and punctuation.

```python
# Benchmark clause with contractions, abbreviations, and monetary punctuation
test_clause = "Alibaba.com isn't liable for user's $500 loss under Sec. 4.2(b), e.g., system outages."

# 1. Python basic split
split_tokens = test_clause.split()

# 2. NLTK word tokenize
nltk_toks = word_tokenize(test_clause)

# 3. spaCy tokenize
spacy_toks = [token.text for token in nlp(test_clause)]

print("=== Tokenisation Comparison on Legal Clause ===")
print("Target Sentence:", test_clause)
print("-" * 70)
print(f"1. Python split()    ({len(split_tokens)} tokens): {split_tokens}")
print(f"2. NLTK word_tokenize ({len(nltk_toks)} tokens): {nltk_toks}")
print(f"3. spaCy doc tokens   ({len(spacy_toks)} tokens): {spacy_toks}")

# Side-by-side observation on tricky tokens
print("\n--- Key Differences in Handling ---")
print("Contraction 'isn\'t':")
print("  -> Python split: ['isn\'t'] (keeps as single raw word with apostrophe)")
print("  -> NLTK:         ['is', 'n\'t'] (separates verb and negation component)")
print("  -> spaCy:        ['is', 'n\'t'] (grammatically splits for syntactic parsing)")
```

### Step 4: Batch Clause Tokenisation on Amazon and Alibaba Corpus
Tokenize the full text files (`amazon.txt` & `alibaba.txt`) and compare corpus-level sentence and token distributions.

```python
# Load full agreements
with open("data/amazon.txt", "r", encoding="utf-8") as f:
    amz_raw = f.read()

with open("data/alibaba.txt", "r", encoding="utf-8") as f:
    ali_raw = f.read()

# Process through spaCy pipeline
doc_amz = nlp(amz_raw)
doc_ali = nlp(ali_raw)

amz_sents = list(doc_amz.sents)
ali_sents = list(doc_ali.sents)

stats_df = pd.DataFrame([
    {
        "Company": "Amazon",
        "Total Sentences": len(amz_sents),
        "Total Tokens": len(doc_amz),
        "Avg Tokens / Sentence": round(len(doc_amz) / len(amz_sents), 2),
        "Total Punctuations": sum(1 for t in doc_amz if t.is_punct),
        "Total StopWords": sum(1 for t in doc_amz if t.is_stop)
    },
    {
        "Company": "Alibaba",
        "Total Sentences": len(ali_sents),
        "Total Tokens": len(doc_ali),
        "Avg Tokens / Sentence": round(len(doc_ali) / len(ali_sents), 2),
        "Total Punctuations": sum(1 for t in doc_ali if t.is_punct),
        "Total StopWords": sum(1 for t in doc_ali if t.is_stop)
    }
])

print("=== Corpus Tokenisation Statistics ===")
display(stats_df)
```

### Step 5: Extension Activity – High-Complexity Legal Sentence Identifier
Identify and rank excessively long or complex legal clauses (by token count and punctuation density) which need summarization most.

```python
# Extension Activity: Find the most complex clauses requiring summarization
clause_complexity = []

for sent in list(doc_amz.sents) + list(doc_ali.sents):
    text = sent.text.strip()
    tokens = [t for t in sent if not t.is_space]
    if len(tokens) > 5:  # filter out section titles
        num_words = sum(1 for t in tokens if t.is_alpha)
        num_punct = sum(1 for t in tokens if t.is_punct)
        clause_complexity.append({
            "Clause_Snippet": text[:90] + "..." if len(text) > 90 else text,
            "Total_Tokens": len(tokens),
            "Word_Count": num_words,
            "Punct_Count": num_punct,
            "Punct_Density_Pct": round((num_punct / len(tokens)) * 100, 2)
        })

df_complex = pd.DataFrame(clause_complexity).sort_values(by="Total_Tokens", ascending=False)

print("=== Top 5 Longest / Most Complex Legal Sentences (Prime Candidates for Summarization) ===")
display(df_complex.head(5))
```
