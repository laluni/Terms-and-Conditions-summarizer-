### 1. Benchmark: NLTK vs spaCy vs Scikit-Learn TF-IDF on Legal Terms
Compare processing speed, token handling, and abstraction capabilities across libraries on contractual sentences.

```python
import time
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample complex legal clause
legal_clause = """
Amazon.com Services LLC and its affiliates provide website features and other products and services. 
To the full extent permissible by law, Amazon disclaims all warranties, express or implied, including merchantability and fitness for a particular purpose. 
Under no circumstances shall the liability exceed $500.00 USD under Section 12.4.
"""

# 1. NLTK Processing
t0 = time.time()
nltk_sents = sent_tokenize(legal_clause.strip())
nltk_words = [word_tokenize(s) for s in nltk_sents]
nltk_time = (time.time() - t0) * 1000

# 2. spaCy Processing
nlp = spacy.load("en_core_web_sm")
t0 = time.time()
spacy_doc = nlp(legal_clause.strip())
spacy_sents = list(spacy_doc.sents)
spacy_ents = [(ent.text, ent.label_) for ent in spacy_doc.ents]
spacy_time = (time.time() - t0) * 1000

# 3. Scikit-Learn TF-IDF Representation
t0 = time.time()
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_mat = vectorizer.fit_transform([legal_clause])
tfidf_features = vectorizer.get_feature_names_out()
tfidf_time = (time.time() - t0) * 1000

print("=== Performance & Capability Comparison on Legal Text ===")
comp_df = pd.DataFrame([
    {"Tool": "NLTK", "Type": "Rule/Statistical (Classical)", "Primary Use in Legal": "Tokenisation & POS Tagging", "Execution Time (ms)": round(nltk_time, 2), "Entities / Output Sample": f"{len(nltk_sents)} Sents, {sum(len(w) for w in nltk_words)} Tokens"},
    {"Tool": "spaCy", "Type": "Industrial Linguistic & Neural", "Primary Use in Legal": "Dependency Parsing & NER", "Execution Time (ms)": round(spacy_time, 2), "Entities / Output Sample": f"{len(spacy_ents)} entities: {spacy_ents[:2]}"},
    {"Tool": "Scikit-Learn / TF-IDF", "Type": "Statistical Vector Space", "Primary Use in Legal": "Salient Term Weighting & Retrieval", "Execution Time (ms)": round(tfidf_time, 2), "Entities / Output Sample": f"Top Terms: {list(tfidf_features[:4])}"}
])
display(comp_df)
```
