### 1. Benchmark: NLTK vs spaCy vs Hugging Face on Legal Terms
Compare processing speed, token handling, and abstraction capabilities across libraries on contractual sentences.

```python
import time
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import spacy
from transformers import pipeline

# Sample complex legal clause
legal_clause = """
Amazon.com Services LLC and its affiliates provide website features and other products and services. 
To the full extent permissible by law, Amazon disclaims all warranties, express or implied, including merchantability and fitness for a particular purpose. 
Under no circumstances shall the liability exceed $500.00 USD under Section 12.4.
"""

# 1. NLTK Processing
t0 = time.time()
nltk_sents = sent_tokenize(legal_clause.strip())
nltk_words = [word_tokenize(s) for s in nltk_sents]
nltk_time = (time.time() - t0) * 1000

# 2. spaCy Processing
nlp = spacy.load("en_core_web_sm")
t0 = time.time()
spacy_doc = nlp(legal_clause.strip())
spacy_sents = list(spacy_doc.sents)
spacy_ents = [(ent.text, ent.label_) for ent in spacy_doc.ents]
spacy_time = (time.time() - t0) * 1000

# 3. Hugging Face Transformers Pipeline (Zero-Shot Classification / Summarization Check)
classifier = pipeline("zero-shot-classification", model="valhalla/distilbart-mnli-12-3")
t0 = time.time()
hf_res = classifier(
    "Under no circumstances shall the liability exceed $500.00 USD under Section 12.4.",
    candidate_labels=["Limitation of Liability", "User Privacy", "Dispute Resolution"]
)
hf_time = (time.time() - t0) * 1000

print("=== Performance & Capability Comparison on Legal Text ===")
comp_df = pd.DataFrame([
    {"Tool": "NLTK", "Type": "Rule/Statistical (Classical)", "Primary Use": "Tokenisation & POS Tagging", "Execution Time (ms)": round(nltk_time, 2), "Entities Detected": "None (Raw Tokenizer)"},
    {"Tool": "spaCy", "Type": "Industrial Linguistic & Neural", "Primary Use": "Dependency Parsing & NER", "Execution Time (ms)": round(spacy_time, 2), "Entities Detected": f"{len(spacy_ents)} entities: {spacy_ents[:2]}"},
    {"Tool": "Hugging Face", "Type": "Pretrained Transformer (DL)", "Primary Use": "Zero-Shot Clause Categorization", "Execution Time (ms)": round(hf_time, 2), "Entities / Top Prediction": f"{hf_res['labels'][0]} ({hf_res['scores'][0]*100:.1f}%)"}
])
display(comp_df)
```
